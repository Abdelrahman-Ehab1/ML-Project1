## KNN Algorithm
will be used for part C and part D

```python
def _find_k_nearest_neighbors(matrix, target , k, similarity_fn ,axis=0 ,get_co_rated_fn =None , sim_cache=None):
    """
    Finds the top k nearest neighbors for a target user or item.

    Args:
        matrix (pd.DataFrame): Utility matrix
        target (int): ID of target user or item
        k (int): Number of neighbors to fetch
        similarity_fn (function): Function to calculate similarity
        axis (int): Axis along which to find neighbors (0 for users, 1 for items)
        get_co_rated_fn (function): Function to get co-rated items/users
        sim_cache (dict): Cache for storing computed similarities

    Returns:
        list[tuple[int, float]]: List of (neighbor_id, similarity_score)
    """
    if get_co_rated_fn is None:
        get_co_rated_fn = get_co_rated

    #if the matrix is precomputed and cached, we can directly fetch the top k neighbors from the cache
    #Ai generated
    if sim_cache is not None and target in sim_cache.index:
        row = sim_cache.loc[target].drop(labels=[target], errors="ignore").dropna()
        return list(row.sort_values(ascending=False).head(k).items())
    #end of generation
    if axis == 0:
        #searching users
        similares = matrix.index
    else:
        #searching items
        similares = matrix.columns
    neighbors=[]
    for similar in similares:
        if similar == target:
            continue

        v1,v2 =get_co_rated_fn(matrix ,target , similar, axis=axis)
        #ai generated
        if len(v1) < 2:
            continue
        #end of generation
        s =similarity_fn(v1,v2)
        if s is None or np.isnan(s):
            continue
        neighbors.append((similar,s))
        #the second element is the sort key and reverse 3la4an teb2a descending order
        neighbors.sort(key= lambda x: x[1], reverse = True)
        return neighbors [:k]
    

        
        
```

### Train/Test Splits [M1]

```python
def train_test_split_ratings(ratings_df, test_fraction=0.2):
    from scipy.sparse import csr_matrix
    user_code = ratings_df['user_id'].astype('category').cat.codes
    movie_code = ratings_df['movie_id'].astype('category').cat.codes
    
    df = ratings_df.copy()
    df['user_code'] = user_code
    df['movie_code'] = movie_code
    
    df = df.sort_values(by=['user_code', 'timestamp']).reset_index(drop=True)
    
    user_counts = df.groupby('user_code').cumcount()
    user_totals = df.groupby('user_code')['user_code'].transform('count')
    
    test_mask = user_counts >= (user_totals * (1 - test_fraction))
    train_mask = ~test_mask
    
    num_users = df['user_code'].max() + 1
    num_movies = df['movie_code'].max() + 1
    
    train_csr = csr_matrix(
        (df.loc[train_mask, 'rating'], (df.loc[train_mask, 'user_code'], df.loc[train_mask, 'movie_code'])),
        shape=(num_users, num_movies)
    )
    
    test_csr = csr_matrix(
        (df.loc[test_mask, 'rating'], (df.loc[test_mask, 'user_code'], df.loc[test_mask, 'movie_code'])),
        shape=(num_users, num_movies)
    )
    
    return train_csr, test_csr
    """
    Splits ratings per-user to guarantee robust testing representations.

    Args:
        ratings_df (pd.DataFrame): Complete interaction DataFrame
        test_fraction (float): Target split proportion
    Returns:
        tuple(pd.DataFrame, pd.DataFrame): train_df, test_df
    """
    # TODO: Member 1 - Implement per-user stratified split
    
```
