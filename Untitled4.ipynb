## 3. Part B — Similarity Metrics [M2]

Member 2 implements vector comparisons, index alignments, and the similarity precomputation cache.

```python
def get_co_rated(matrix, a, b, axis=0):
    if axis == 0:
        # Compare rows (users). a and b are user IDs.
        vec_a = matrix.loc[a]
        vec_b = matrix.loc[b]
    elif axis == 1:
        # Compare columns (items). a and b are movie IDs.
        vec_a = matrix[a]
        vec_b = matrix[b]
    else:
        raise ValueError("axis must be 0 (users/rows) or 1 (items/columns)")

    mask = vec_a.notna() & vec_b.notna()

    # Apply mask and convert to NumPy arrays
    return vec_a[mask].to_numpy(), vec_b[mask].to_numpy()

def euclidean_similarity(vec1, vec2):
    if len(vec1) < 2:
        return np.nan
    d = np.sqrt(np.sum((vec1 - vec2) ** 2))
    return 1.0 / (1.0 + d)

def cosine_similarity(vec1, vec2):
    if len(vec1) < 2:
        return np.nan
    norm1 = np.linalg.norm(vec1)
    norm2 = np.linalg.norm(vec2)
    if norm1 == 0 or norm2 == 0:
        return np.nan
    return np.dot(vec1, vec2) / (norm1 * norm2)

def pearson_similarity(vec1, vec2):
    if len(vec1) < 2:
        return np.nan
    v1_centered = vec1 - np.mean(vec1)
    v2_centered = vec2 - np.mean(vec2)
    return cosine_similarity(v1_centered, v2_centered)

# ===== BEGIN AI OPTIMIZATION (Claude) =====
# Faster "one target vs. everyone" similarity.
# It does the same job as the loop in _find_k_nearest_neighbors (get_co_rated -> similarity_fn),
# but with plain NumPy instead of pandas, and it calls YOUR similarity functions,
# so the numbers are exactly the same as before:
#   1. Turn the matrix into a NumPy array (NaN = missing) and a True/False "was rated?" array, once.
#   2. For a target, one NumPy comparison tells which other rows/columns share >= 2 rated entries with it.
#   3. Call similarity_fn only on those; every other pair is undefined (NaN) anyway.
_CACHE = {"matrix": None, "arrays": {}, "neighbors": {}}   # remembers work done for ONE matrix

def clear_cache():
    """Forget everything cached. Call this if you edit a matrix in place."""
    _CACHE.update(matrix=None, arrays={}, neighbors={})

def _cache_for(matrix):
    # The cache belongs to one matrix; if a different matrix is passed in, start fresh.
    if _CACHE["matrix"] is not matrix:
        _CACHE.update(matrix=matrix, arrays={}, neighbors={})
    return _CACHE

def _similarities_to_all(matrix, target, similarity_fn, axis=0):
    """Similarity of `target` to EVERY row (axis=0) or column (axis=1), as a NumPy array
    (NaN = undefined). Same result as get_co_rated + similarity_fn for each pair."""
    arrays = _cache_for(matrix)["arrays"]
    if axis not in arrays:
        values = matrix.to_numpy(dtype=float)
        if axis == 1:
            values = np.ascontiguousarray(values.T)      # make the items the rows
        arrays[axis] = (values, ~np.isnan(values))       # (ratings, "was rated?")
    values, rated = arrays[axis]
    labels = matrix.index if axis == 0 else matrix.columns
    t = labels.get_loc(target)                           # raises KeyError for an unknown target
    both = rated & rated[t]                              # True where the candidate AND the target rated
    enough = np.where(np.count_nonzero(both, axis=1) >= 2)[0]   # >= 2 co-rated, otherwise undefined
    sims = np.full(len(labels), np.nan)
    target_ratings = values[t]
    for i in enough:
        mask = both[i]
        sims[i] = similarity_fn(target_ratings[mask], values[i][mask])
    return sims
# ===== END AI OPTIMIZATION (Claude) =====

def precompute_similarity_matrix(matrix, similarity_fn, axis=0):
    if axis == 0:
        labels = matrix.index       # user IDs
    else:
        labels = matrix.columns     # movie IDs

    # ===== BEGIN AI OPTIMIZATION (Claude) =====
    # Same table as the loop below: one row per label, using the faster helper above.
    # (The original pairwise loop below is kept for reference; it is no longer reached.)
    table = np.array([_similarities_to_all(matrix, a, similarity_fn, axis=0 if axis == 0 else 1) for a in labels])
    return pd.DataFrame(table, index=labels, columns=labels)
    # ===== END AI OPTIMIZATION (Claude) =====

    sim_df = pd.DataFrame(np.nan, index=labels, columns=labels)

    for i, a in enumerate(labels):
        for b in labels[i:]:
            vec1, vec2 = get_co_rated(matrix, a, b, axis=axis)
            score = similarity_fn(vec1, vec2)
            sim_df.loc[a, b] = score
            sim_df.loc[b, a] = score

    return sim_df
```

## KNN Algorithm
will be used for part C and part D

```python
# ===== BEGIN AI OPTIMIZATION (Claude) =====
# Neighbour cache. The ranked neighbour list of a target depends only on
# (matrix, metric, axis, target) -- NOT on k or on the item being predicted. So it is
# computed once per target and every k / every test row reuses it
# (top-k is just the first k entries of the same sorted list, so the results are identical).
_FAST_TOP = 50   # neighbours stored per target by default (largest k in the assignment grid)

def _compute_neighbors(matrix, targets, similarity_fn, axis, top):
    store = _cache_for(matrix)["neighbors"]
    labels = matrix.index if axis == 0 else matrix.columns
    for target in targets:
        sims = _similarities_to_all(matrix, target, similarity_fn, axis)
        sims[labels.get_loc(target)] = np.nan                       # a target is never its own neighbour
        pairs = [(labels[i], sims[i]) for i in np.where(~np.isnan(sims))[0]]   # drop undefined scores
        pairs.sort(key=lambda x: x[1], reverse=True)                # same sort as the original loop
        store[(axis, similarity_fn, target)] = (pairs[:top], len(pairs) <= top)   # (list, is it the full list?)

def _fast_neighbors(matrix, target, k, similarity_fn, axis):
    """Top-k [(neighbour_id, score), ...] from the cache (computed on first use)."""
    axis = 0 if axis == 0 else 1
    store = _cache_for(matrix)["neighbors"]
    key = (axis, similarity_fn, target)
    entry = store.get(key)
    if entry is None or not (entry[1] or k <= len(entry[0])):       # not cached, or cached list too short for this k
        _compute_neighbors(matrix, [target], similarity_fn, axis, max(k, _FAST_TOP))
        entry = store[key]
    return entry[0][:k]

def _prefetch_neighbors(matrix, targets, similarity_fn, axis, top=_FAST_TOP):
    """Compute the neighbour lists for many targets up-front."""
    labels = matrix.index if axis == 0 else matrix.columns
    store = _cache_for(matrix)["neighbors"]
    targets = pd.unique(np.asarray(targets))
    targets = targets[pd.Index(targets).isin(labels)]       # unknown targets are left to fail/skip as before
    todo = [t for t in targets if (axis, similarity_fn, t) not in store]
    _compute_neighbors(matrix, todo, similarity_fn, axis, top)

def _prefetch_for_test(matrix, test_df, similarity_fn, method, top=_FAST_TOP):
    if method == "user":
        _prefetch_neighbors(matrix, test_df["userId"].to_numpy(), similarity_fn, 0, top)
    else:
        _prefetch_neighbors(matrix, test_df["movieId"].to_numpy(), similarity_fn, 1, top)
# ===== END AI OPTIMIZATION (Claude) =====


def _find_k_nearest_neighbors(matrix, target , k, similarity_fn ,axis=0 ,get_co_rated_fn =None , sim_cache=None):
    """
    Finds the top k nearest neighbors for a target user or item.

    Args:
        matrix (pd.DataFrame): Utility matrix
        target (int): ID of target user or item
        k (int): Number of neighbors to fetch
        similarity_fn (function): Function to calculate similarity
        axis (int): Axis along which to find neighbors (0 for users, 1 for items)
        get_co_rated_fn (function): Function to get co-rated items/users
        sim_cache (dict): Cache for storing computed similarities

    Returns:
        list[tuple[int, float]]: List of (neighbor_id, similarity_score)
    """
    if get_co_rated_fn is None:
        get_co_rated_fn = get_co_rated

    #if the matrix is precomputed and cached, we can directly fetch the top k neighbors from the cache
    #Ai generated
    if sim_cache is not None and target in sim_cache.index:
        row = sim_cache.loc[target].drop(labels=[target], errors="ignore").dropna()
        return list(row.sort_values(ascending=False).head(k).items())
    #end of generation
    # ===== BEGIN AI OPTIMIZATION (Claude) =====
    # Fast path: default co-rated function -> cached neighbour list built with NumPy.
    # A custom get_co_rated_fn uses the original loop below.
    if get_co_rated_fn is get_co_rated:
        return _fast_neighbors(matrix, target, k, similarity_fn, axis)
    # ===== END AI OPTIMIZATION (Claude) =====
    if axis == 0:
        #searching users
        similares = matrix.index
    else:
        #searching items
        similares = matrix.columns
    neighbors=[]
    for similar in similares:
        if similar == target:
            continue

        v1,v2 =get_co_rated_fn(matrix ,target , similar, axis=axis)
        #ai generated
        if len(v1) < 2:
            continue
        #end of generation
        s =similarity_fn(v1,v2)
        if s is None or np.isnan(s):
            continue
        neighbors.append((similar,s))
        #the second element is the sort key and reverse 3la4an teb2a descending order
    neighbors.sort(key= lambda x: x[1], reverse = True)
    return neighbors [:k]
    

        
        
```

## 4. Part C — User-based CF [M3]

Member 3 implements user neighbor isolation and user-based prediction strategies.

```python
def find_k_nearest_users(matrix, target_user, k, similarity_fn, get_co_rated_fn=None, sim_cache=None)-> list[ tuple[int , float]]: 

    return _find_k_nearest_neighbors(matrix, target_user ,k , similarity_fn , axis =0, get_co_rated_fn=get_co_rated_fn, sim_cache=sim_cache)

def predict_rating_user_based(matrix, target_user, target_item, k, similarity_fn, get_co_rated_fn=None, sim_cache=None):
    """
    Predicts rating for an item by an user based on neighbor user ratings.

    Args:
        matrix (pd.DataFrame): Utility matrix
        target_user (int): ID of target user
        target_item (int): ID of target movie/item
        k (int): Number of neighbors
        similarity_fn (function): Similarity metric
    Returns:
        float or None: Predicted rating, or None if prediction is unfeasible
    """
    #AI generated
    if target_item not in matrix.columns:
        return None # if the target item is not in the matrix, we cannot predict its rating
    #end of generation
    neighbors = find_k_nearest_users(matrix ,target_user , k, similarity_fn,get_co_rated_fn=get_co_rated_fn, sim_cache=sim_cache)
    # ===== BEGIN AI OPTIMIZATION (Claude) =====
    item_ratings = matrix[target_item]   # single column lookup, reused for all k neighbours
    # ===== END AI OPTIMIZATION (Claude) =====
    num =0.0
    denom= 0.0
    for neighbor, sim_score in neighbors:
        # ===== BEGIN AI OPTIMIZATION (Claude) =====
        # original: rating = matrix.loc[neighbor, target_item]
        rating = item_ratings.at[neighbor]
        # ===== END AI OPTIMIZATION (Claude) =====
        if pd.isna(rating):
            continue
        num= num + sim_score*rating
        denom+= abs(sim_score) #AI added abs() to avoid negative weights
    return None if denom == 0.0 else num / denom
```

## 5. Part D — Item-based CF [M3]

Member 3 implements item-level recommendations and item-based prediction mechanisms.

```python
def find_k_nearest_items(matrix, target_item, k, similarity_fn, get_co_rated_fn=None, sim_cache=None):

   return _find_k_nearest_neighbors(matrix,target_item,k, similarity_fn ,axis = 1, get_co_rated_fn=get_co_rated_fn, sim_cache=sim_cache)

def predict_rating_item_based(matrix, target_user, target_item, k, similarity_fn,
                              get_co_rated_fn=None, sim_cache=None):
    """
    Predicts rating for an item by a user based on item similarities.

    Returns:
        float or None: Predicted rating, or None if prediction is unfeasible
    """
    if target_user not in matrix.index:
        return None

    neighbors = find_k_nearest_items(matrix, target_item, k, similarity_fn,
                                     get_co_rated_fn=get_co_rated_fn,
                                     sim_cache=sim_cache)

    # ===== BEGIN AI OPTIMIZATION (Claude) =====
    user_ratings = matrix.loc[target_user]   # single row lookup, reused for all k neighbours
    # ===== END AI OPTIMIZATION (Claude) =====
    num, denom = 0.0, 0.0
    for neighbor_item, sim_score in neighbors:
        # ===== BEGIN AI OPTIMIZATION (Claude) =====
        # original: rating = matrix.loc[target_user, neighbor_item]
        rating = user_ratings.at[neighbor_item]
        # ===== END AI OPTIMIZATION (Claude) =====
        if pd.isna(rating):
            continue
        num += sim_score * rating
        denom += abs(sim_score)

    return None if denom == 0.0 else num / denom
```

```python
def evaluate(matrix, test_df, k, similarity_fn, method, get_co_rated_fn=None, sim_cache=None):
    """
    Scores ratings predictions on test partition using RMSE, MAE and reports un-predicted fractions.

    Args:
        matrix (pd.DataFrame): Training matrix
        test_df (pd.DataFrame): Test interaction DataFrame
        k (int): Neighbors
        similarity_fn (function): Distance metric
        method (str): 'user' or 'item' prediction process selection
    Returns:
        dict: {'rmse': float, 'mae': float, 'none_frac': float}
    """
    #Ai generated
    assert method in ("user", "item"), "method must be 'user' or 'item'"
    #fails if method not  "user" nor "item"
    #end of generation
    # ===== BEGIN AI OPTIMIZATION (Claude) =====
    # Compute every needed neighbour list up-front in blocks (only when the fast path applies).
    if sim_cache is None and (get_co_rated_fn is None or get_co_rated_fn is get_co_rated):
        _prefetch_for_test(matrix, test_df, similarity_fn, method, max(k, _FAST_TOP))
    # ===== END AI OPTIMIZATION (Claude) =====
    errors = []
    none_count = 0
    # ===== BEGIN AI OPTIMIZATION (Claude) =====
    # original:
    # for _, row in test_df.iterrows():
    #     u, i, true_r = int(row["userId"]), int(row["movieId"]), float(row["rating"])
    # iterrows builds a pandas Series per row (slow); iterate over plain arrays instead.
    for u, i, true_r in zip(test_df["userId"].to_numpy(), test_df["movieId"].to_numpy(), test_df["rating"].to_numpy()):
        u, i, true_r = int(u), int(i), float(true_r)
    # ===== END AI OPTIMIZATION (Claude) =====

        if method == "user":
            pred = predict_rating_user_based(matrix, u, i, k, similarity_fn,
                                             get_co_rated_fn=get_co_rated_fn,
                                             sim_cache=sim_cache)
        else:
            pred = predict_rating_item_based(matrix, u, i, k, similarity_fn,
                                             get_co_rated_fn=get_co_rated_fn,
                                             sim_cache=sim_cache)
        if pred is None:
            none_count += 1
        else:
            errors.append(true_r - pred)

    errors = np.array(errors)
    n = len(errors)
    return {
        "rmse": float(np.sqrt(np.mean(errors ** 2))) if n else float("nan"),
        "mae":  float(np.mean(np.abs(errors))) if n else float("nan"),
        "n_predictions": n,
        "none_frac": none_count / len(test_df) if len(test_df) else float("nan"),
    }
```

### Q18: Complexity and Scalability Analysis [M1]

#### 1. Time Complexity
* **Similarity Matrix Precomputation:** Computing pairwise similarities across all users requires $O(U^2 \cdot M)$ in the worst case, where $U$ is the total number of users and $M$ is the total number of items. For each user pair, time is proportional to the number of co-rated items.
* **Inference / Prediction Time:** For every test interaction, searching for the top-$k$ nearest neighbors takes $O(U)$ time. Across a test set of size $T$, the total evaluation grid search time scales as $O(\text{Configs} \cdot T \cdot U)$, where $\text{Configs}$ accounts for the combination of similarity metrics and $k$ values.

#### 2. Space Complexity
* **Utility Matrix Storage:** A dense user-item matrix requires $O(U \cdot M)$ memory. Utilizing sparse matrix representations (such as SciPy's `csr_matrix`) reduces space complexity to $O(N)$, where $N$ is the total number of non-zero rating interactions.
* **Similarity Cache:** Storing a precomputed user-user similarity matrix requires $O(U^2)$ memory, which can become a major memory bottleneck as the user base scales.

#### 3. Scalability Bottlenecks
* **User Growth Vulnerability:** User-based collaborative filtering scales poorly when the number of users ($U$) vastly exceeds the number of items ($M$). As platforms grow, maintaining and updating a $U \times U$ similarity matrix becomes computationally prohibitive.
* **Sparsity Impact:** Real-world datasets exhibit high sparsity (often exceeding $99\%$ empty entries). High sparsity reduces the overlap of co-rated items between users, resulting in undefined similarity scores ($\text{NaN}$) and increasing the unpredicted fraction (`none_frac`) during evaluation.

#### 4. Mitigation & Architectural Optimizations
* **Caching Mechanisms:** Precomputing and caching similarity scores prevents redundant distance calculations during hyperparameter grid searches.
* **Item-Based Alternative:** In systems where users outnumber items ($U > M$), item-based collaborative filtering is often preferred because item relationships are more stable and item spaces change less frequently than user behavior.
